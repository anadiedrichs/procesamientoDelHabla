# Trabajando con documentos de texto



## Descargar un archivo PDF

```python
# prompt: descargar el pdf alojado en https://www.comisionporlamemoria.org/wp-content/plugins/pdfjs-viewer-shortcode/pdfjs/web/viewer.php?file=https://www.comisionporlamemoria.org/archivos/materiales/Nunca-Mas.pdf&attachment_id=0&dButton=true&pButton=true&oButton=false&sButton=true&pagemode=none&_wpnonce=2860df8d99

import requests

def download_pdf(url, filename):
  """Downloads a PDF file from a given URL.

  Args:
    url: The URL of the PDF file.
    filename: The name of the file to save the PDF as.
  """
  try:
    response = requests.get(url, stream=True)
    response.raise_for_status()  # Raise an exception for bad status codes (e.g., 404)

    with open(filename, 'wb') as file:
      for chunk in response.iter_content(chunk_size=8192):
        file.write(chunk)
    print(f"PDF downloaded successfully to {filename}")

  except requests.exceptions.RequestException as e:
    print(f"Error downloading PDF: {e}")

# Example usage
pdf_url = "https://www.comisionporlamemoria.org/wp-content/plugins/pdfjs-viewer-shortcode/pdfjs/web/viewer.php?file=https://www.comisionporlamemoria.org/archivos/materiales/Nunca-Mas.pdf&attachment_id=0&dButton=true&pButton=true&oButton=false&sButton=true&pagemode=none&_wpnonce=2860df8d99"
output_filename = "Nunca-Mas.pdf"

download_pdf(pdf_url, output_filename)
```
